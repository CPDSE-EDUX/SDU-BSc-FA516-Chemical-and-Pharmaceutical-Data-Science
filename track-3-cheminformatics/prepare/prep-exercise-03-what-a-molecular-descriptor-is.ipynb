### Task 5 — Predict, then check

Glucose and fructose are different sugars — an aldose and a ketose — with the same formula.
So they should have the same molecular weight.

Write down what you expect these lines to print, in particular the last one, before you run anything:

```py
glucose = Chem.MolFromSmiles("OCC(O)C(O)C(O)C(O)C=O")
fructose = Chem.MolFromSmiles("OCC(=O)C(O)C(O)C(O)CO")
print("glucose MW:", round(Descriptors.MolWt(glucose), 4))
print("fructose MW:", round(Descriptors.MolWt(fructose), 4))
print("same value:", round(Descriptors.MolWt(glucose),4) == round(Descriptors.MolWt(fructose), 4))
```

Now run them and compare with your prediction.
