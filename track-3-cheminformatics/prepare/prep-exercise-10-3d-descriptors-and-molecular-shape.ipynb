## Exercise — 3D descriptors and molecular shape

### Task 1 — Which one needs a conformer?

You are given salicylic acid and two descriptor calls:

```py
Descriptors.MolWt(mol)
Descriptors3D.RadiusOfGyration(mol)
```

One of these works on a molecule straight from `MolFromSmiles`; the other
fails or is meaningless, because a SMILES carries no coordinates.

Parse `"OC(=O)c1ccccc1O"`, check the parse, and print only the descriptor
that a molecule without a conformer can supply, rounded to 2 decimals.


### Task 2 — Look at the structures first

Here are three compounds with quite different shapes:

```py
smiles = {
    "paracetamol": "CC(=O)Nc1ccc(O)cc1",
    "celecoxib": "Cc1ccc(cc1)-c1cc(nn1c1ccc(cc1)S(N)(=O)=O)C(F)(F)F",
    "salicylic acid": "OC(=O)c1ccccc1O",
}
```

Parse all three, checking each parse, and draw them side by side in one row
with `Draw.MolsToGridImage`, using the names as legends.


### Task 3 — Build the conformers

For each of the three molecules, follow the workflow from the video:
add explicit hydrogens with `Chem.AddHs`, embed with
`AllChem.EmbedMolecule(..., randomSeed=42)` and relax with
`AllChem.MMFFOptimizeMolecule`.

Store the hydrogen-added molecules in a dictionary `mols_h` keyed by name,
and print for each one the heavy-atom count and the count with hydrogens.


### Task 4 — Asphericity and radius of gyration

For each conformer in `mols_h`, print the asphericity to 3 decimals and the
radius of gyration to 2 decimals, using
`Descriptors3D.Asphericity` and `Descriptors3D.RadiusOfGyration`.

Recall from the video that asphericity is $0$ for a sphere and $1$ for a rod,
and that a large radius of gyration means an extended molecule.


### Task 5 — Flatness against aromatic content

For each conformer, print the plane-of-best-fit value from
`rdMolDescriptors.CalcPBF` to 2 decimals together with the number of aromatic
rings from `rdMolDescriptors.CalcNumAromaticRings`.

Before you run it, predict which of the three has the **lowest** PBF, then
check whether your prediction matches the printed numbers.


### Task 6 — The same molecule, three seeds

Celecoxib has two freely rotating aryl links, so it can take on very
different shapes. Embed a fresh hydrogen-added celecoxib three times, once
for each of the seeds `1`, `7` and `42`, optimising after each embedding, and
print the radius of gyration to 2 decimals for each seed.

Predict first whether the three numbers will be identical. Then state, from
the output, which two things you would have to record in a paper for someone
else to reproduce your value.
