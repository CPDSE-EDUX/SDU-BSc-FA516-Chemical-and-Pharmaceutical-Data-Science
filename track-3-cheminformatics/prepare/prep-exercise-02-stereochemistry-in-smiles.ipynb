## Exercise — Stereochemistry in SMILES

### Task 1 — Same connectivity, different drug

Ibuprofen is sold as a racemate, but only the S-enantiomer inhibits COX.
The two enantiomers are:

```py
"CC(C)Cc1ccc(cc1)[C@@H](C)C(=O)O"   # S-ibuprofen
"CC(C)Cc1ccc(cc1)[C@H](C)C(=O)O"    # R-ibuprofen
```

Parse both, check neither is `None`, and print whether their canonical SMILES
(from `Chem.MolToSmiles`, track 1 video 8) are equal. Then print whether their
canonical SMILES written with `isomericSmiles=False` are equal. The two answers
must differ — that difference is exactly the pharmacology the flat string loses.


### Task 2 — Predict, then check: which strings describe the same molecule?

Look at these four SMILES and decide, before running anything, what each of the

```py
a = Chem.MolFromSmiles("C[C@@H](N)C(=O)O")
b = Chem.MolFromSmiles("C[C@H](N)C(=O)O")
c = Chem.MolFromSmiles("OC(=O)/C=C/C(=O)O")
d = Chem.MolFromSmiles("OC(=O)/C=C\\C(=O)O")

print(Chem.MolToSmiles(a) == Chem.MolToSmiles(b))
print(Chem.MolToSmiles(c) == Chem.MolToSmiles(d))
print(Chem.MolToSmiles(c) == Chem.MolToSmiles(Chem.MolFromSmiles("O/C(=O)... ")))
```

Ignore the third line — instead predict and then check these two lines only:
the alanine comparison and the butenedioic acid comparison. Write your two

### Task 3 — Find the stereocentres

```py
"C[C@H](c1ccc2cc(OC)ccc2c1)C(=O)O"   # naproxen, assigned
"CC(c1ccc2cc(OC)ccc2c1)C(=O)O"       # naproxen, no marker
```

Parse both and print `Chem.FindMolChiralCenters(mol, includeUnassigned=True)`
for each, labelled `"assigned:"` and `"unassigned:"`. Say nothing in prose —
the `?` in the second output is the point.
