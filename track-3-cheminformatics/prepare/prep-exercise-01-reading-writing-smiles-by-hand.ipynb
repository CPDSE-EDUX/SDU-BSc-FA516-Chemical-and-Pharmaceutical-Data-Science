### Task 3 — Aromatic and Kekulé

Pyridine can be written in lowercase aromatic form as `c1ccncc1`, or in Kekulé form with alternating single and double bonds as `C1=CC=NC=C1`.

Write down what you expect the following three lines to print, then run the block and compare.

```py
pyridine_arom = Chem.MolFromSmiles("c1ccncc1")
pyridine_kek = Chem.MolFromSmiles("C1=CC=NC=C1")
print(Chem.MolToSmiles(pyridine_arom))
print(Chem.MolToSmiles(pyridine_kek))
print(Chem.MolToSmiles(pyridine_arom) == Chem.MolToSmiles(pyridine_kek))


### Task 4 — Inside the brackets

Write three SMILES by hand, each using square brackets, and remember the order inside them: isotope, symbol, hydrogens, charge.

| species | what to write |
|---|---|
| hydroxide ion | an oxygen with one explicit hydrogen and one negative charge |
| methylammonium | a methyl carbon bonded to a nitrogen with three explicit hydrogens and one positive charge |
| carbon-13 labelled methanol | methanol whose carbon is the isotope 13 |

Parse each one, and print the canonical SMILES of all three.


### Task 5 — Fix the paracetamol string

A classmate wrote paracetamol by hand as follows, and reports that nothing works afterwards:

```py
paracetamol = Chem.MolFromSmiles("CC(=O)Nc1ccc(O)cc")
print(paracetamol.GetNumAtoms())
```
