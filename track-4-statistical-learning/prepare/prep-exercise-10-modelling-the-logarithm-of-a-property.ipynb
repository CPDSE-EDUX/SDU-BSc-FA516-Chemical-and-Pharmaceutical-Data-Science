## Exercise — Modelling the logarithm of a property

### Task 1 — Spot the long right tail

Build the demonstration set from the video with exactly these lines:

```py
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

np.random.seed(42)
logp = np.random.normal(2.5, 1.2, 60)
sol_uM = 10 ** (3.0 - 0.8 * logp + np.random.normal(0, 0.3, 60))
df = pd.DataFrame({"logp": logp, "sol_uM": sol_uM})
```

Then do two things:

- Print how many orders of magnitude the solubility spans. This is $\log_{10}$ of the largest value divided by the smallest, so use `np.log10`, `np.max` and `np.min` and round to 1 decimal.
- Draw a histogram of `df["sol_uM"]` with 20 bins. Give it axis labels with units and a title.

Check that the histogram shows the shape the video warned about: most compounds crowd into the first bars, and a few stretch the axis to the right.


````
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

np.random.seed(42)
logp = np.random.normal(2.5, 1.2, 60)
sol_uM = 10 ** (3.0 - 0.8 * logp + np.random.normal(0, 0.3, 60))
df = pd.DataFrame({"logp": logp, "sol_uM": sol_uM})
print("orders of magnitude spanned:", round(np.log10(np.max(df["sol_uM"]) / np.min(df["sol_uM"])), 1))
plt.figure(figsize=(8, 6))
plt.hist(df["sol_uM"], bins=20)
plt.xlabel("Solubility (µM)")
plt.ylabel("Number of compounds")
plt.title("Raw solubility: a long right tail")
plt.grid(True)
plt.show()
````

### Task 2 — Refit on the log scale and compare the residuals

Use the same straight-line model as the video:

```py
def line(x, a, b):
    return a * x + b
```

1. Fit `line` to the raw `df["sol_uM"]` against `df["logp"]` with `curve_fit`. Unpack the parameters into `a_raw, b_raw` and store the residuals (observed minus predicted) as `resid_raw`.
2. Add a column `df["log_sol"]` holding `np.log10` of the solubility. Fit the same `line` to it, unpack into `a_log, b_log`, store the residuals as `resid_log` and print the slope rounded to 2 decimals.
3. Put a number on the funnel. Take the compounds with `logP < 2.5` and those with `logP >= 2.5`. For each model, print the standard deviation (`np.std`) of the residuals in each of those two groups.
4. Draw the residual plot of the log model, with a black zero line from logP −1 to 6, axis labels with units and a title.

Before running, predict which model gives roughly equal spreads in the two logP groups. Then compare your residual plot with the raw-scale residual plot from the video, which used this same data.


````
def line(x, a, b):
    return a * x + b

popt, pcov = curve_fit(line, df["logp"], df["sol_uM"])
a_raw, b_raw = popt
resid_raw = df["sol_uM"] - line(df["logp"], a_raw, b_raw)
df["log_sol"] = np.log10(df["sol_uM"])
popt, pcov = curve_fit(line, df["logp"], df["log_sol"])
a_log, b_log = popt
resid_log = df["log_sol"] - line(df["logp"], a_log, b_log)
print("log-model slope:", round(a_log, 2))
low = df["logp"] < 2.5
high = df["logp"] >= 2.5
print("raw residual SD (uM), low / high logP:", round(np.std(resid_raw[low]), 1), round(np.std(resid_raw[high]), 1))
print("log residual SD (log uM), low / high logP:", round(np.std(resid_log[low]), 2), round(np.std(resid_log[high]), 2))
plt.figure(figsize=(8, 6))
plt.scatter(df["logp"], resid_log)
plt.plot([-1, 6], [0, 0], color="black")
plt.xlabel("logP")
plt.ylabel("Residual (log µM)")
plt.title("Residuals: straight line on log solubility")
plt.grid(True)
plt.show()
````

### Task 3 — Fix the error band

A colleague predicts the solubility of a new compound with logP 1.5, using the log model from Task 2. Their code is:

```py
pred_log_new = line(1.5, a_log, b_log)
resid_sd = np.std(resid_log)
sol_new_uM = 10 ** pred_log_new
low_uM = sol_new_uM - resid_sd
high_uM = sol_new_uM + resid_sd
```

The prediction is correct, but the range comes out implausibly narrow, less than one micromolar wide, and perfectly symmetric. The cause is that `resid_sd` is in log units and has been added directly to a value in µM.

Correct the two bound lines so they go one standard deviation down and up **on the log scale** before converting back. Then print:

- the predicted solubility in µM
- the range in µM
- the distance below and the distance above the prediction

All values should be rounded to 2 decimals.


````
pred_log_new = line(1.5, a_log, b_log)
resid_sd = np.std(resid_log)
sol_new_uM = 10 ** pred_log_new
low_uM = 10 ** (pred_log_new - resid_sd)
high_uM = 10 ** (pred_log_new + resid_sd)
print("predicted solubility (uM):", round(sol_new_uM, 2))
print("range (uM):", round(low_uM, 2), "to", round(high_uM, 2))
print("below:", round(sol_new_uM - low_uM, 2), "above:", round(high_uM - sol_new_uM, 2))
````
