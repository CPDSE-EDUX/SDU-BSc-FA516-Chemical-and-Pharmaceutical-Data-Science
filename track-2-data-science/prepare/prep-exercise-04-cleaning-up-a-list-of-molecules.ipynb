## Exercise — Cleaning up a list of molecules

You are given a small, deliberately messy bioactivity table.

```py
import pandas as pd
from rdkit import Chem
from rdkit.Chem.MolStandardize import rdMolStandardize
raw = pd.DataFrame({
    "name": ["paracetamol", "ibuprofen", "salbutamol sulfate", "paracetamol (repeat)", "junk entry"],
    "smiles": ["CC(=O)Nc1ccc(O)cc1", "CC(C)Cc1ccc(cc1)C(C)C(=O)O", "CC(C)(C)NCC(O)c1ccc(O)c(CO)c1.OS(=O)(=O)O", "Oc1ccc(NC(C)=O)cc1", "c1ccccc1((("],
    "pIC50": [5.0, 4.5, 7.2, 5.4, 6.1]})
```

### Task 3 — Predict, then check: are these the same compound?

The two paracetamol records in the table are spelled differently. Write down what each of these four lines prints **before** you run anything.

```py
mol_a = Chem.MolFromSmiles("CC(=O)Nc1ccc(O)cc1")
mol_b = Chem.MolFromSmiles("Oc1ccc(NC(C)=O)cc1")
print("raw comparison:", "CC(=O)Nc1ccc(O)cc1" == "Oc1ccc(NC(C)=O)cc1")
print("canonical comparison:", Chem.MolToSmiles(mol_a) == Chem.MolToSmiles(mol_b))
```
