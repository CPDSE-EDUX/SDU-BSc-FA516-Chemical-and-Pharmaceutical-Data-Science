## Exercise — Exploring a chemical dataset step by step

You are given a small table of eight compounds with a measured aqueous solubility.

```py
data = {"name": ["aspirin", "ibuprofen", "paracetamol", "caffeine",
                 "metformin", "warfarin", "diazepam", "naproxen"],
        "mw": [180.16, 206.29, 151.16, 194.19, 129.16, 308.33, 284.74, 230.26],
        "logp": [1.19, 3.97, 0.46, -0.07, -1.43, 2.70, 2.82, 3.18],
        "hbd": [1, 1, 2, 0, 5, 1, 0, 1],
        "bbb_class": ["no", "yes", "yes", "yes", "no", "yes", "yes", "no"],
        "solubility_mg_L": [3300.0, 21.0, 14000.0, None, 1380000.0, 17.0, 50.0, None]}
```

Build this table as a DataFrame called `sol` and run the first four steps of the

### Task 3 — Predict, then check

Aqueous solubility in mg/L runs from a few mg/L for a lipophilic drug to over a
gram per litre for metformin. Before running anything, write down on paper
whether you expect the mean of `solubility_mg_L` to be far **above** or far
**below** its median.

Then run exactly these lines and compare with your prediction:

```py
solubility = sol["solubility_mg_L"].dropna()
print("mean (mg/L):", round(solubility.mean(), 1))
print("median (mg/L):", round(solubility.median(), 1))
print("min:", solubility.min(), "max:", solubility.max())
```
