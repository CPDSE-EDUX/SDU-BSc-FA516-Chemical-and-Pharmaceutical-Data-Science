### Task 2 — Predict: text versus structure

The SMILES of caffeine is `Cn1cnc2c1c(=O)n(C)c(=O)n2C`. Look at the lines below and
**write down what each `print` will show before you run anything**. Pay attention to
which line counts characters and which counts atoms.

```py
caffeine_text = "Cn1cnc2c1c(=O)n(C)c(=O)n2C"
print("characters in the string:", len(caffeine_text))

from rdkit import Chem
caffeine = Chem.MolFromSmiles(caffeine_text)
if caffeine is None:
    print("invalid SMILES:", caffeine_text)
print("heavy atoms:", caffeine.GetNumAtoms())
print("bonds:", caffeine.GetNumBonds())
```

Now run them and compare with your prediction.
