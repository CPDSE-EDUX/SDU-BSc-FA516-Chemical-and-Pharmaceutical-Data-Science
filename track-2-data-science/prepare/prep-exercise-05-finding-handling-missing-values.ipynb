## Exercise — Finding & handling missing values

You are given a small table of six drugs with three measured properties. Some
entries were never recorded.

```py
import pandas as pd
import numpy as np

assays = pd.DataFrame({
    "name": ["aspirin", "caffeine", "paracetamol", "naproxen", "loratadine", "celecoxib"],
    "mw": [180.2, 194.2, 151.2, 230.3, 382.9, 381.4],
    "logp": [1.19, -0.07, 0.46, np.nan, 5.20, np.nan],
    "tpsa": [63.6, 58.4, 49.3, 46.5, np.nan, 77.9],
    "ic50_nm": [3200.0, np.nan, np.nan, np.nan, np.nan, 40.0]
})
```

### Task 1 — Count the gaps

Create this table, then print the number of missing values per column, and
print the missing **fraction** per column rounded to two decimals.


### Task 2 — Predict the shapes

The table has 6 rows and 5 columns. Write down, on paper, what each of these
three lines prints — before you run anything.

```py
print(assays.dropna().shape)
print(assays.dropna(how="all").shape)
print(assays.dropna(thresh=4).shape)
```

Then run exactly these three lines and compare with your prediction.


### Task 3 — Fill logP sensibly

Print the mean and the median of the `logp` column, both rounded to two
decimals. Loratadine's logP of 5.20 pulls the mean upwards, so this column is
skewed by an outlier.

Then work on a **copy** of the table called `filled`, fill the `logp` gaps with
whichever of the two the video recommended for skewed data, and print the
`name` and `logp` columns of `filled`.


### Task 4 — Clean the table

`ic50_nm` has only two real measurements out of six, while `tpsa` is missing
just one value. Apply the rule of thumb from the video:

- drop the sparse column with `dropna` on `axis=1` and a threshold of 4
  non-missing values,
- impute the single `tpsa` gap with that column's median.

Start from `filled` (whose logP gaps are already imputed), call the result
`clean`, print it, and print the missing-value count per column to confirm
nothing is left.
