## Exercise — Looping with pairs & positions

You are given three parallel lists describing four antibiotics.

```py
antibiotics = ["amoxicillin", "doxycycline", "azithromycin", "ciprofloxacin"]
dose_mg = [500, 100, 250, 500]
mw = [365.40, 444.43, 748.98, 331.34]
```

### Task 1 — Name and dose in step

Copy the three lists, then loop over the names and the doses together with `zip()`, unpacking each turn into `name` and `dose`. Print one line per drug in the form `amoxicillin dose: 500 mg`.


### Task 2 — Dose in millimoles

Now walk all three lists together in one loop, unpacking into `name`, `dose` and `weight`. For each drug compute the dose in millimoles as $n = \dfrac{\text{dose in mg}}{\text{MW}}$ and print a line like

```
amoxicillin: 500 mg = 1.37 mmol
```

using an f-string with two decimal places for the millimole value.


### Task 3 — Predict, then check

Write down on paper exactly what these two loops print, before you run anything.

```py
for i, name in enumerate(antibiotics):
    print(i, name)

for i, name in enumerate(antibiotics, start=1):
    print(i, name)
```

Then run them and compare with your prediction. How many lines does each loop print, and what is the first number in each?


### Task 4 — Choose zip over range(len(...))

This loop is meant to print each name with its molecular weight, but it indexes two lists by hand and the brackets have been mixed up, so every line reports the wrong weight:

```py
for i in range(len(antibiotics)):
    print(antibiotics[i], mw[i - 1])
```

Rewrite it as a single `zip()` loop with unpacked names, so that no index appears at all. Print one line per drug in the form `amoxicillin MW: 365.4`.
