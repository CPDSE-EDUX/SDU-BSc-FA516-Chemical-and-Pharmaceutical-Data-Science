### Task 4 — Ibuprofen as a record

The `molecular_weight` function from the video is:

```py
def molecular_weight(formula, masses):
    total = 0.0
    for element, count in formula.items():
        total = total + count * masses[element]
    return total
```

Define it, then build a dictionary `ibuprofen` with the keys `"name"` (`"ibuprofen"`), `"dose_mg"` (`400`) and `"formula"` (the formula $\mathrm{C_{13}H_{18}O_2}$ as a dictionary).

Print the dose, the number of carbon atoms, and the molecular weight rounded to 2 decimals.
