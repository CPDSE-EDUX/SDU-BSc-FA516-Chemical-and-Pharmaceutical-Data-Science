## Exercise — Drawing & visualising molecules

### Task 1 — Draw salicylic acid and save it

Import `Chem`, `Draw`, `AllChem`, `rdMolDraw2D` and `os`, then build salicylic acid from the SMILES `OC(=O)c1ccccc1O`.

Compute its 2D coordinates, make a `300 x 300` image with `Draw.MolToImage`, and save it as `salicylic.png`.

Print the number of conformers the molecule has after computing coordinates, the image size, and the size of the file on disk in bytes.


### Task 2 — A grid of three analgesics

You are given this dictionary:

```py
analgesics = {"Salicylic acid": "OC(=O)c1ccccc1O",
              "Paracetamol": "CC(=O)Nc1ccc(O)cc1",
              "Naproxen": "COc1ccc2cc(ccc2c1)C(C)C(=O)O"}
```

Turn the SMILES into molecules with a list comprehension and the names into a list of legends. Draw them with `Draw.MolsToGridImage`, three per row, sub-image size `(250, 200)`, legends on, and `useSVG=False`.

Write the result to `analgesics_grid.png` and print the file size in bytes.


### Task 3 — Predict, then annotate

Look at these two lines, using the `salicylic` molecule from Task 1 (SMILES `OC(=O)c1ccccc1O`):

```py
pattern = Chem.MolFromSmarts("C(=O)[OX2H1]")
match = salicylic.GetSubstructMatch(pattern)
print(match)
```

Write down which atom indices you expect before running anything. Remember that atom 0 is the first atom in the SMILES.

Then run those lines, and draw salicylic acid on a `400 x 300` Cairo canvas with atom indices switched on **and** the matched atoms highlighted in red, `(1.0, 0.4, 0.4)`. Save it as `salicylic_annotated.png` and print the matched indices.
