## Exercise — Variables, types & operators

### Task 1 — Four variables for one prescription

A prescription is for ibuprofen, 20 tablets of 0.400 g each, and the drug is not soluble in water.

Create four variables:

- `n_tablets` — the whole number 20
- `mass_g` — the decimal number 0.400
- `compound` — the text `"ibuprofen"`
- `soluble` — the true-or-false value `False`

Print all four on one line. Then the pharmacist corrects the count to 30 tablets: reassign `n_tablets` and print it again.


### Task 2 — A day of ibuprofen

A patient may take at most 1200 mg of ibuprofen per day, and the tablets contain 400 mg each.

Assign `daily_max_mg = 1200` and `dose_mg = 400`, then compute and print, each on its own line:

- the sum `daily_max_mg + dose_mg`
- the difference `daily_max_mg - dose_mg`
- the product `dose_mg * 3`
- the division `daily_max_mg / 1000` (the daily maximum in grams)
- the floor division `daily_max_mg // dose_mg` (whole tablets that fit)
- the modulo `daily_max_mg % dose_mg` (milligrams left over)
- `10 ** 3` (the number of milligrams in a gram)


### Task 3 — Predict, then check

Write down on paper what each of these five lines prints, **before** you run anything. Pay attention to which ones say `int` and which say `float`.

```py
print(type(dose_mg))
print(type(mass_g))
print(type(compound))
print(type(soluble))
print(type(daily_max_mg / 1000))
```

Now run them and compare with your predictions.

State to yourself which error Python raises and why. Then write one line that prints the compound name and the dose on the same line, without any `+` between them — the video's very first `print` with several comma-separated values does exactly this.
