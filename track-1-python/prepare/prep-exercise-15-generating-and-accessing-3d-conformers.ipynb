### Task 3 — Predict: embedding without hydrogens

The lines below embed paracetamol **without** calling `AddHs` first, then print
the number of atoms carrying coordinates and the number of hydrogens still
stored implicitly on atom 0.

```py
mol_noh = Chem.MolFromSmiles("CC(=O)Nc1ccc(O)cc1")
AllChem.EmbedMolecule(mol_noh, randomSeed=42)
conf_noh = mol_noh.GetConformer()
print(conf_noh.GetNumAtoms())
print(mol_noh.GetAtomWithIdx(0).GetTotalNumHs())
```

Write down both numbers before you run anything, and say which atoms of the
molecule have no position at all. Then run the block and compare.
