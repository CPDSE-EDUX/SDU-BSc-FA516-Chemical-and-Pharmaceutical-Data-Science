## Exercise — List comprehensions & generators

### Task 1 — Rewrite the loop

You are given four molecular weights in g/mol:

```py
weights = [151.16, 180.16, 206.28, 230.26]
```

The loop below builds a list of the number of millimoles in one gram of each substance, rounded to three decimals:

```py
mmol_per_g = []
for w in weights:
    mmol_per_g.append(round(1000 / w, 3))
```

Rewrite it as a single list comprehension called `mmol_per_g` and print the result.


### Task 2 — Keep only the heavy ones

Using the same `weights` list, build a list called `heavy_mmol` containing the millimoles per gram (rounded to three decimals) **only** for compounds with a molecular weight above 180 g/mol.

Print `heavy_mmol`. It must contain two values.

### Task 4 — Generator, once only

Read the following lines and write down on paper what each `print` produces, before running anything:

```py
g = (round(1000 / w, 3) for w in weights)
print("total:", round(sum(g), 3))
print("again:", sum(g))
print("direct:", round(sum(1000 / w for w in weights), 3))
```

Then run exactly these lines and compare with your prediction. Pay attention to the second line.
