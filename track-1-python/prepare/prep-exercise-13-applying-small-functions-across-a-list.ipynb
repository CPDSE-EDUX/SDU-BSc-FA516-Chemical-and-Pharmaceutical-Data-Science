## Exercise — Applying small functions across a list

### Task 1 — A lambda for tablet strength

A tablet contains a mass of active ingredient in milligrams, and you want it expressed in grams.

Write a lambda that takes one number in milligrams and returns it divided by 1000. Store it in a variable called `mg_to_g`, then call it on the 500 mg aspirin dose and print the result.


### Task 2 — Convert a whole list at once

Here are the molecular weights in g/mol of four compounds, in the same order as in the video:

```py
mw = [180.16, 151.16, 206.28, 194.19]
```

Use `map()` with your `mg_to_g` function is *not* what is wanted here — instead use `map()` with a new lambda that multiplies each molecular weight by 1000, wrap the result in `list()`, store it in `mg_per_mol` and print it.


### Task 3 — Predict, then check

Look at these three lines and write down, on paper, exactly what each `print` shows — including how many elements each list has.

```py
print(list(filter(lambda x: x < 200, mw)))
print(list(map(lambda x: x < 200, mw)))
print(len(list(filter(lambda x: x < 200, mw))))
```

Then run them and compare with your prediction.
