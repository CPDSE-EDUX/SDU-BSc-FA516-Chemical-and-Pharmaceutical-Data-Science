### Task 4 — Fix the formulary update

A colleague wants to update this formulary list, but their code is wrong.

```py
formulary = ["Aspirin", "Ibuprofen", "Paracetamol"]
formulary.append("Codeine", "Morphine")
formulary.remove(0)
```

`append` takes only one item, and `remove` deletes by **value**, not by position — so both lines fail.

Rewrite the update so that `"Codeine"` and `"Morphine"` are both added to the end, `"Aspirin"` is deleted by name, and the item now at position `0` is replaced by `"Ibuprofen sodium"`. Print the list at the end.
