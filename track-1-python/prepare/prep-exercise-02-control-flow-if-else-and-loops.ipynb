## Exercise — Control flow: if/else and loops

### Task 1 — Classify a buffer

A buffer solution has been measured at `pH = 7.0`.

Assign that value to a variable `pH`, then write an `if` / `elif` / `else` chain
that prints `"acidic"` when the pH is below 7, `"neutral"` when it is exactly 7,
and `"basic"` otherwise.


### Task 2 — Masses of ethanol

The molar mass of ethanol is $46.07$ g/mol.

Store it in a variable `M_ethanol`, then use a `for` loop with `range()` to print
the mass in grams of 1, 2, 3, 4 and 5 moles. Print the number of moles and the
mass on each line, as in the video's water example.


### Task 3 — Predict the comparisons

A tablet contains a dose of 500 mg. Write down, on paper, what each of these five
lines prints, before you run anything.

```py
dose_mg = 500
print(dose_mg > 500)
print(dose_mg == 500)
print(dose_mg != 250)
print(dose_mg <= 500)
print(dose_mg < 500)
```

Then run the block and compare it with your prediction.


### Task 4 — Stop at the first bad reading

The following code should report only the **first** reading above 8 and then stop.
Instead it prints every reading above 8.

```py
readings = [6.8, 7.4, 5.1, 8.2, 7.0, 9.1]
for pH in readings:
    if pH > 8:
        print("found basic reading:", pH)
```

Write the corrected version so that only `found basic reading: 8.2` is printed.
